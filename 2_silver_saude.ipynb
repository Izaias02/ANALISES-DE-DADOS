```r
# ============================================================
# Camada Silver: Download, Limpeza e Tipagem (Saúde via API)
# ============================================================
library(readr)
library(googledrive)

# 1. Mapeando os caminhos da Aula 6 conforme configurado no Bloco 2
pasta_aula    <- "Aula 6 - 31/08/2026"
pasta_origem  <- paste0(pasta_aula, "/0 - Bases de Dados/5 - Saúde")
nome_arquivo  <- "hospitais_leitos_sp_2025_tratado.csv"
caminho_completo_base <- paste0(pasta_origem, "/", nome_arquivo)

cat("🔍 Buscando o arquivo original no Google Drive...\n")

# Localiza o arquivo bruto no Drive usando a API
arquivo_encontrado <- drive_get(path = caminho_completo_base)

# Faz o download limpo para a sessão atual do Colab
drive_download(
  file = arquivo_encontrado,
  path = "/content/hospitais_leitos_sp_2025_tratado.csv",
  overwrite = TRUE
)

# 2. Lendo o arquivo baixado para a memória do R
df_hospitais <- read_csv(
  "/content/hospitais_leitos_sp_2025_tratado.csv",
  locale = locale(encoding = "UTF-8"),
  show_col_types = FALSE
)

# 3. Aplicando a Regra de Purificação da Silver (Removendo Duplicações)
df_hospitais_silver <- df_hospitais[df_hospitais$registro_unico_para_analise == "Sim", ]

# ============================================================
# 4. EXECUÇÃO DO ROTEIRO DE INSPEÇÃO DA ATIVIDADE
# ============================================================

# Visualiza o dataframe tratado.
df_hospitais_silver

# Exibe os primeiros registros limpos.
head(df_hospitais_silver, 10)

# Quantidade de linhas e colunas (Demonstra o antes e depois da limpeza!).
cat("\n📊 Dimensões da Base Silver (Linhas x Colunas):\n")
print(dim(df_hospitais_silver))

# Nome de todas as colunas disponíveis.
cat("\n📋 Nomes das Colunas:\n")
print(names(df_hospitais_silver))

# Estrutura e tipagem das variáveis pós-tratamento.
cat("\n🔬 Estrutura das Variáveis:\n")
str(df_hospitais_silver)
```

```r
# ============================================================
# 4. FILTRAGEM DE DUPLICIDADES E HIGIENIZAÇÃO (SILVER) - VERSÃO BASE
# ============================================================

# Filtragem direta usando colchetes [linhas, colunas] - Método padrão e estável no R
df_hospitais_silver <- df_hospitais[df_hospitais$registro_unico_para_analise == "Sim", ]

# Remove possíveis linhas que venham completamente vazias (NA) por segurança
df_hospitais_silver <- df_hospitais_silver[!is.na(df_hospitais_silver$registro_unico_para_analise), ]

# ============================================================
# 5. DIAGNÓSTICO DO ANTES E DEPOIS (ROTEIRO OFICIAL)
# ============================================================

# Visualiza os primeiros registros filtrados
head(df_hospitais_silver, 10)

# Quantidade de linhas e colunas pós-limpeza (Compare com o anterior!)
cat("\n📊 Dimensões da Base após a limpeza na Camada Silver:\n")
print(dim(df_hospitais_silver))

# Nome de todas as colunas que restaram e estão higienizadas
cat("\n📋 Colunas prontas para o SBC:\n")
print(names(df_hospitais_silver))

# Estrutura final e conferência de tipos das variáveis
cat("\n🔬 Estrutura técnica pós-tratamento:\n")
str(df_hospitais_silver)
```

```r
# ============================================================
# 6. EXPORTAÇÃO DA CAMADA SILVER (MÉTODO ID DIRETO - INFALÍVEL)
# ============================================================
library(googledrive)

cat("🔍 Buscando identificador único da pasta '2-silver' na nuvem...\n")

# Faz uma busca global no Drive pelo nome da pasta para capturar o ID correto de rede
busca_pasta_silver <- drive_find(pattern = "2-silver", type = "folder", n_max = 1)

if (nrow(busca_pasta_silver) > 0) {
  cat("✅ Pasta encontrada! ID de Rede:", busca_pasta_silver$id[1], "\n")

  # Salva o arquivo limpo temporariamente no espaço do Colab (/content)
  write.csv(df_hospitais_silver, "/content/hospitais_leitos_sp_2025_silver.csv", row.names = FALSE)

  cat("📤 Iniciando upload do arquivo tratado utilizando o ID seguro...\n")

  # Faz o upload usando a linha da tabela localizada pela busca, ignorando caminhos de texto
  drive_upload(
    media = "/content/hospitais_leitos_sp_2025_silver.csv",
    path = as_id(busca_pasta_silver$id[1]),
    name = "hospitais_leitos_sp_2025_silver.csv",
    overwrite = TRUE
  )

  cat("🎉 SUCESSO ABSOLUTO, IZAIAS! O R quebrou o bloqueio e salvou a Camada Silver no seu Drive!\n")

} else {
  cat("❌ ERRO: A pasta '2-silver' não foi localizada na busca global.\n")
  cat("👉 Como plano de emergência, execute esta linha para salvar o arquivo na tela local do seu Colab:\n")
  write.csv(df_hospitais_silver, "/content/hospitais_leitos_sp_2025_silver.csv", row.names = FALSE)
  cat("O arquivo 'hospitais_leitos_sp_2025_silver.csv' aparecerá na barra da esquerda para baixar no seu PC!\n")
}
```
