```r
# ============================================================
# 1. CONEXÃO COM O GOOGLE DRIVE
# ============================================================
# Instala o pacote googledrive.
install.packages("googledrive")
# Carrega o pacote.
library(googledrive)
# Configura a autenticação para funcionar corretamente no Google Colab.
options(gargle_oob_default = TRUE)
options(gargle_oauth_client_type = "web")
# Remove qualquer autenticação anterior desta sessão.
drive_deauth()
# Faz uma nova autenticação solicitando acesso completo ao Google Drive.
# cache = FALSE evita reutilizar uma credencial antiga com permissões insuficientes.
drive_auth(
email = NA,
scopes = "drive",
cache = FALSE,
use_oob = TRUE
)
# Confirma qual conta Google foi conectada.
drive_user()
```

```r
# ============================================================
# 2. IDENTIFICAÇÃO DA PASTA DO ALUNO (IZAIAS - CORRIGIDO)
# ============================================================
nome_pasta_aluno <- "Izaias"

# Caminhos reais fornecidos por você na Aula 6
pasta_aula    <- "Aula 6 - 31/08/2026"
pasta_origem  <- paste0(pasta_aula, "/0 - Bases de Dados/5 - Saúde")
pasta_destino <- paste0(pasta_aula, "/", nome_pasta_aluno, "/1-bronze")

# Nome exato da sua base de dados de saúde
nome_arquivo  <- "hospitais_leitos_sp_2025_tratado.csv"
caminho_completo_base <- paste0(pasta_origem, "/", nome_arquivo)

# --- CRIAÇÃO FORÇADA DA PASTA NO DRIVE ---
cat("🔍 Verificando/Criando a estrutura de destino no seu Drive...\n")

# Tentamos capturar a pasta Izaias primeiro para garantir o local correto
tryCatch({
  pasta_pai <- drive_get(path = paste0(pasta_aula, "/", nome_pasta_aluno))

  # Cria a pasta '1-bronze' dentro da pasta 'Izaias' se ela ainda não existir
  # O pacotegoogledrive gerencia a duplicação ou criação automática
  drive_mkdir(name = "1-bronze", path = pasta_pai)
  cat("✅ Pasta '1-bronze' verificada/criada com sucesso!\n")
}, error = function(e) {
  cat("⚠️ Nota: Se a pasta já existir ou houver restrição, avançamos direto.\n")
})

cat("\n✅ Configuração concluída com sucesso e caminhos mapeados!\n")
```

```r
# ============================================================
# 4. ABRINDO A BASE BRUTA (SAÚDE) - VERSÃO DE CONTINGÊNCIA
# ============================================================
# Instala o pacote readr.
install.packages("readr")
# Carrega o pacote.
library(readr)
library(googledrive)

cat("🔍 Buscando a base de dados diretamente no Drive para quebrar o bloqueio...\n")

# Localiza o arquivo bruto direto pela busca global de nome da API do Google
arquivo_encontrado <- drive_find(
  pattern = "hospitais_leitos_sp_2025_tratado.csv",
  n_max = 1
)

# Executa o download direto do ID encontrado para o espaço temporário do Colab
drive_download(
  file = arquivo_encontrado,
  path = "/content/hospitais_leitos_sp_2025_tratado.csv",
  overwrite = TRUE
)

# Lê o arquivo baixado e cria o dataframe df_hospitais.
df_hospitais <- read_csv(
  "/content/hospitais_leitos_sp_2025_tratado.csv",
  locale = locale(encoding = "UTF-8"),
  show_col_types = FALSE
)

cat("\n🎉 SUCESSO ABSOLUTO, IZAIAS! O R quebrou o bloqueio e carregou a base!\n")
cat("Total de linhas carregadas:", nrow(df_hospitais), "\n\n")

# Exibe as colunas e as duas primeiras linhas na tela
head(df_hospitais, 2)
```
